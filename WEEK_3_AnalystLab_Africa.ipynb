# **SUPERVISED AND UNSUPERVISED LEARNING**

## Supervised Learning

Supervised learning is a type of machine learning where a model learns from labelled data. This means that the dataset contains both input variables (features) and a known output variable (target).

The aim of supervised learning is to learn the relationship between the features and the target variable so that predictions can be made on new, unseen data.

In this Titanic dataset, supervised learning is applied because the target variable, **Survived**, is already known.

The model uses passenger characteristics such as:
- Passenger class
- Age
- Fare
- Other available passenger information

to learn patterns associated with survival outcomes.

Examples of supervised learning algorithms include:
- Logistic Regression
- Decision Trees
- Random Forest
- Support Vector Machines

Real-world applications of supervised learning include:
- Predicting customer credit risk
- Email spam classification
- Medical diagnosis prediction
- Fraud detection

## Unsupervised Learning

Unsupervised learning is a type of machine learning where a model learns patterns from data that does not have labelled outcomes.

The goal is to discover hidden structures, relationships, or groups within the dataset without being given a specific answer to predict.

Examples of unsupervised learning algorithms include:
- K-Means Clustering
- Hierarchical Clustering
- Principal Component Analysis (PCA)

Real-world applications of unsupervised learning include:
- Customer segmentation
- Grouping similar products
- Detecting unusual patterns in data

Unlike supervised learning, unsupervised learning does not predict a known target variable. Instead, it explores the underlying structure of the data.

# **OVERFITTING AND UNDERFITTING**

When building a machine learning model, the goal is not only to perform well on the data it was trained on, but also to make accurate predictions on new, unseen data.

# OVERFITTING

Overfitting occurs when a machine learning model learns the training data too closely, including noise and random patterns that do not represent the true relationship in the data.

As a result, the model performs very well on the training data but performs poorly on unseen testing data.

In the Titanic dataset, an overfitted model may memorize specific patterns from the training passengers instead of learning general relationships between passenger characteristics and survival outcomes.

For example, a model may learn very specific combinations of passenger features that only exist in the training data but fail to predict survival accurately for new passengers.

# EFFECTS OF OVERFITTING

- High accuracy on training data.
- Poor performance on testing data.
- Reduced ability to generalize to new observations.


# REDUCING OVERFITTING

- Use more training data.
- Reduce unnecessary features.
- Use simpler models.
- Apply regularization techniques.
- Use cross-validation to evaluate model performance.

## UNDERFITTING

Underfitting occurs when a machine learning model is too simple to capture important patterns in the data.

The model performs poorly on both the training data and testing data because it has not learned enough information from the dataset.

In the Titanic dataset, an underfitted model may fail to capture important relationships between passenger characteristics such as age, passenger class, fare, and survival.

#EFFECTS OF UNDERFITTING

- Poor performance on training data.
- Poor performance on testing data.
- The model cannot identify important patterns.

# REDUCING UNDERFITTING

- Use a more complex model.
- Include more relevant features.
- Improve feature selection.
- Train the model for longer where applicable.


A good machine learning model should find the balance between overfitting and underfitting.

The aim is to create a model that learns meaningful patterns from the Titanic dataset and can accurately predict survival outcomes for passengers it has not seen before.




## **Model Evaluation Basics**

## Evaluating Machine Learning Models

After a machine learning model has been trained, it is important to evaluate how well it performs. Model evaluation helps determine whether the model has learned useful patterns and whether it can make reliable predictions on unseen data.

For a classification problem such as Titanic survival prediction, evaluation metrics are used to compare the model's predicted outcomes with the actual survival outcomes.

The actual values are the recorded survival outcomes in the dataset, while the predicted values are the outcomes generated by the machine learning model.

## Accuracy

Accuracy measures the proportion of correct predictions made by the model out of all predictions.

The formula is:

Accuracy = (Correct Predictions) / (Total Predictions)

A higher accuracy value indicates that the model correctly predicted a larger percentage of observations.

However, accuracy alone may not always provide a complete picture, especially when classes are imbalanced.

For the Titanic dataset, accuracy represents the percentage of passengers for whom the model correctly predicted whether they survived or did not survive.


## Confusion Matrix

A confusion matrix is a table used to evaluate the performance of a classification model.

It compares the actual outcomes with the predicted outcomes.

The confusion matrix contains four values:

- **True Positive (TP):** The model correctly predicted a passenger survived.
- **True Negative (TN):** The model correctly predicted a passenger did not survive.
- **False Positive (FP):** The model predicted survival, but the passenger did not survive.
- **False Negative (FN):** The model predicted that a passenger did not survive, but the passenger actually survived.

The confusion matrix helps identify the types of errors made by the model.

## Precision

Precision measures how many of the passengers predicted as survivors actually survived.

Formula:

Precision = True Positives / (True Positives + False Positives)

A high precision value means that when the model predicts survival, it is usually correct.

## Recall

Recall measures how many of the actual survivors were correctly identified by the model.

Formula:

Recall = True Positives / (True Positives + False Negatives)

A high recall value means the model successfully identifies most passengers who survived.

## Determining Model Performance

A model should not be evaluated using only one metric. Different evaluation measures provide different information about model performance.

For the Titanic dataset:
- Accuracy shows the overall percentage of correct predictions.
- The confusion matrix shows the types of prediction errors.
- Precision shows how reliable survival predictions are.
- Recall shows how well the model identifies actual survivors.

A model performs well when it achieves good results on unseen testing data and maintains a balance between correctly identifying survivors and avoiding incorrect predictions.

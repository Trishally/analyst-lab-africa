# **ADVANCED MACHINE LEARNING**

This project explores advanced machine learning algorithms for predicting passenger survival on the Titanic dataset.The objective is to compare different classification models identify the best-performing model based on evaluation metrics.The model implemented include Decision Tree, Random Forest and Gradient Boosting.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report,f1_score, precision_score, recall_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
```

# **FEATURE SELECTION**

```python
x=df.drop("Survived",axis=1)
y=df["Survived"]
```

# **TRAIN-TEST SPLIT**

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

# **DECISION TREE CLASSIFICATION MODEL**

A decision tree is a supervised machine learning model used for classification and regression tasks.It predicts outcomes by learning a series of decision rules from the training data.I'll use a decision tree to predict wether a passenger survived or died.

```python
dt_model=DecisionTreeClassifier()
dt_model.fit(x_train,y_train)
dt_pred=dt_model.predict(x_test)
```

# **PREDICTIONS**


```python
#Predicting on the test data
dt_pred = dt_model.predict(x_test)

print ("Accuracy:", accuracy_score(y_test, dt_pred))
print ("F1 Score:", f1_score(y_test, dt_pred))
print ("Recall:", recall_score(y_test, dt_pred))
print ("Precision:", precision_score(y_test, dt_pred))
```

```python
dt_accuracy = accuracy_score(y_test, dt_pred)
dt_f1 = f1_score(y_test, dt_pred)
dt_recall = recall_score(y_test, dt_pred)
dt_precision = precision_score(y_test, dt_pred)

dt_results=pd.DataFrame({
    "Model": ['Decision Tree'],
    "Accuracy": [dt_accuracy],
    "F1 Score": [dt_f1],
    "Recall": [dt_recall],
    "Precision": [dt_precision]
})
dt_results
```

# **RANDOM FOREST MODEL**

Random Forest is an ensemble machine learning algorithm that combines multiple decision trees to make predictions.It uses several trees and aggregates their results,which can improve predictive accuracy and reduce overfittting.

```python
rf_model= RandomForestClassifier()
rf_model.fit(x_train,y_train)
rf_pred=rf_model.predict(x_test)
```

```python
rf_accuracy=accuracy_score(y_test,rf_pred)
rf_f1=f1_score(y_test,rf_pred)
rf_recall=recall_score(y_test,rf_pred)
rf_precision=precision_score(y_test,rf_pred)

print("Accuracy:",rf_accuracy)
print("F1 Score:",rf_f1)
print("Recall:",rf_recall)
print("Precision:",rf_precision)
```

```python
rf_results= pd.DataFrame({'Model': ['Random Forest'],
                          'Accuracy': [rf_accuracy],
                          'F1 Score': [rf_f1],
                          'Recall': [rf_recall],
                          'Precision': [rf_precision]})
rf_results
```

# **RANDOM FOREST MODEL PERFORMANCE**


The Random Forest model was trained using the Titanic dataset and evaluated using Accuracy, Precision, Recall, and F1-Score.

The model achieved an accuracy of approximately 83.79%, indicating that it correctly classified the majority of passenger survival outcomes in the testing dataset. The model achieved a precision score of 84.62%, meaning that most passengers predicted as survivors were correctly classified.

The recall score of 74.32% shows that the model was able to identify a significant proportion of actual survivors. The F1-Score of 79.14% demonstrates a balance between precision and recall.

The Random Forest model showed improved performance compared to the Decision Tree model, demonstrating the advantage of combining multiple decision trees to improve predictive performance and reduce overfitting.

# **GRADIENT BOOSTING MODEL**

Gradient Boosting is an ensemble machine learning model that combines multiple week learners, usually decision trees, to create a stronger predictive model.Gradient boosting builds trees sequentially, with each new tree focusing on improving the errors made by previous trees.

```python
gb_model=GradientBoostingClassifier(random_state=42)
gb_model.fit(x_train,y_train)
gb_pred=gb_model.predict(x_test)
```

```python
gb_accuracy=accuracy_score(y_test,gb_pred)
gb_precision=precision_score(y_test,gb_pred)
gb_recall=recall_score(y_test,gb_pred)
gb_f1=f1_score(y_test,gb_pred)

print("Accuracy:",gb_accuracy)
print("Precision:",gb_precision)
print("Recall:",gb_recall)
print("F1 Score:",gb_f1)
```

```python
gb_results= pd.DataFrame({'Model': ['Gradient Boosting'],
                          'Accuracy': [gb_accuracy],
                          'F1 Score': [gb_f1],
                          'Recall': [gb_recall],
                          'Precision': [gb_precision]})
gb_results
```

```python
lr_results= pd.DataFrame({'Model': ['Logistic Regression'],
                          'Accuracy': [lr_accuracy],
                          'F1 Score': [lr_f1],
                          'Recall': [lr_recall],
                          'Precision': [lr_precision]})
lr_results
```

# **HYPERPARAMETER TUNING**

Hyperparameter tuning is the process of finding the best combination of model parametersto improve predictive performance.Grid search and cross validation will be used to identify the optimal hyperparameters for the Random Forest Classifier.

```python
from sklearn.model_selection import GridSearchCV

#Define the parameter grid
param_grid={
    'n_estimators':[50,100,200],
    'max_depth':[None,5,10],
    'min_samples_split':[2,5,10],}

#Create the grid search model
grid_search=GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    scoring='accuracy'
)

#Fit the model
grid_search.fit(x_train,y_train)

#Get the best parameters
best_params=grid_search.best_params_
print("Best Parameters:",best_params)
```

```python
print("Best Cross-Validation Accuracy:",grid_search.best_score_)
```

# **EVALUATION OF THE TUNED MODEL**



```python
best_rf=grid_search.best_estimator_
best_rf_pred=best_rf.predict(x_test)

best_accuracy=accuracy_score(y_test,best_rf_pred)
best_precision=precision_score(y_test,best_rf_pred)
best_recall=recall_score(y_test,best_rf_pred)
best_f1=f1_score(y_test,best_rf_pred)

print("Accuracy:",best_accuracy)
print("Precision:",best_precision)
print("Recall:",best_recall)
print("F1 Score:",best_f1)
```

# **HYPERPARAMETER TUNING RESULTS **

Hyperparameter tuning was performed on the Random Forest model using Grid Search with 5-fold cross-validation. The objective was to identify the combination of hyperparameters that produced the best predictive performance.

The optimal hyperparameters were:

Maximum Depth: None (no limit on tree depth)
Minimum Samples Split: 10
Number of Estimators: 200

A maximum depth of None allows each decision tree to grow without a predefined depth limit, while a minimum sample split of 10 helps reduce overfitting by requiring at least ten samples before a node can be split. Using 200 decision trees improved the stability and predictive performance of the model.

The tuned Random Forest model achieved a best cross-validation accuracy of 83.14%, indicating strong and consistent performance across the validation folds.

Although the tuned model achieved a cross validation accuracy of approximately 83.14%,its performance on the test dataset was slightly lower than the default random forest.This suggests that the default random forest model generalized slightly better to the unseen test data.

Therefore the original Random Forest model was retained as the final model because it achieved the highest test accuracy.

# **FEARTURE SELECTION**

The dataset is divided into input features (x) and a target variable (y).The features represent the passenger characteristics used by the model to learn patterns and make predictictions.

**Features (x):** Passenger information used to predict survival.

**Target (y):**The survived variable which indicates wether a passenger survived or not.

```python
x=df.drop('Survived',axis=1)
y=df['Survived']
```

```python
x.head()
```

```python
y.head()
```

# **TRAIN-TEST SPLIT**

The dataset is divided into training and testing sets before building the machine learning model.80% of the data is used for training and 20% is reserved for testing .A random state is specified to ensure that the same split is produced every time the coode is executed.

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

# **TRAIN THE LOGISTIC REGRESSION MODEL**

The logistics regression is a supervised machine learning model used for classification problems.In this project we will classify wether a passenger survived the Titanic disasteror not.The model learns patterns from the training data by identifying relationships between passenger characteristics and survival outcomes. The trained model will then be used to classify passengersin the testing dataset.

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

# **PREDICTIONS**

**0:**Did not Survive

**1:**Survived


```python
y_pred = model.predict(x_test)
print(y_pred)
```

```python
comparison = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})
print(comparison)
```

# **MODEL EVALUATION**

The performance of the Logistic Regression model will be evaluated using the Accuracy Score. The accuracy measures the proportion of correctly classified abservations out of the total number of observations.A higher accuracy score indicates that the model correctly predicts passenger survival more often.The accuracy score provides an overall measure of the model's classification performance on the testing dataset.

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

# **CONCLUSION**

The Logistic Regression model achieved an accuracy score of approximately 80.44%, meaning it correctly predicted the survival outcome for about 80.44% of the passengers in the testing dataset. This indicates that the model was able to identify meaningful patterns in the passenger data and make reliable predictions.
